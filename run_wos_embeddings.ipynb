# Generate WoS Document Embeddings

This notebook embeds documents from WoS partitions using `nomic-ai/nomic-embed-text-v1.5`.

Outputs are saved to `data/embeddings/{partition_name}/` with:
- `embeddings.npz` - Compressed embeddings array
- `metadata.parquet` - Metadata with UT IDs and text info
- `meta.json` - Processing progress tracking

**Configure partitions and settings below, then run all cells.**

```python
from wos_embeddings import process_partitions, EmbedConfig

partitions = ["2"]
batch_size = 8  # Number of texts to embed per batch 
resume = True  # Skip already-processed records

config = EmbedConfig(
    batch_size=batch_size,
    resume=resume,
)
```

```python
config
```

```python
# Process partitions
results = process_partitions(partitions, config)
```

Each result entry shows the processing summary for a partition with status, counts, and timing.

```python
# Display results summary
results
```
